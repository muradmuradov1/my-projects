```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Vüsal", "Günel", None, "Elvin", "Aysu", "Kamran"],
    "age": [28, np.nan, 35, 22, np.nan, 41],
    "salary": [1500, 2200, 1800, np.nan, 1200, 3000],
    "department": ["IT", "HR", "IT", "Sales", "HR", None]
})


df = df.dropna(subset=["name"])          
df["age"] = df["age"].fillna(df["age"].mean()) 
df["salary"] = df["salary"].fillna(0)
df["department"] = df["department"].fillna("Naməlum")

def categorize(age_val):
  
  if age_val < 30:
    return "genc"
  else:
    return "tecrubeli"
df["age_category"] = df["age"].apply(categorize)


salary_descending = df.sort_values(by='salary', ascending=False)


print(df)
print(salary_descending)
```
