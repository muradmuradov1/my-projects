```python
import pandas as pd

df = pd.DataFrame({ 
                  "student": ["Ali","Vəli","Səbinə","Kamran","Aygün","Elvin","Nərmin","Tural"], 
                  "subject": ["Math","Math","Physics","Math","Physics","Chemistry","Chemistry","Physics"], 
                   "score": [85, 92, 78, 88, 95, 70, 65, 82] 
                  
                  })



avg_score = df.groupby("subject")["score"].mean()

student_count = df.groupby("subject")["student"].size() 
student_count = df["student"].value_counts()

scores = df.groupby("subject")["score"].agg(["min", "max"])

high_scores = df[df["score"] > 90]
high_scorers_by_subject = high_scores["subject"].value_counts()

print(avg_score)
print(student_count)
print(scores)
print(high_scorers_by_subject)

```
