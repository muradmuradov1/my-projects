```python
import pandas as pd
df = pd.read_csv("netflix.csv")
# Ensure 'release_year' is numeric for statistical operations
df['release_year'] = pd.to_numeric(df['release_year'], errors='coerce')

def movie():
  while True:
    customer = input("===== NETFLIX DATASET ANALYZER =====\n1. Dataset Information\n2. Statistics\n3. Search Movie\n4. Movies by Year\n5. Top Genres\n6. Top Directors\n7. Country Analysis\n8. Missing Values\n9. Exit\nEnter a section: ")
    if customer == "1":
      print(df.info())
      print(df.columns)
      print(df.shape)
    elif customer == "2": # Compare with string "2"
      statistics = input("1. Oldest Movie\n2. Newest movie\n3. Average release year\n4. Number of Movies\n5. Number of TV Shows: ")
      if statistics == "1":
        print(f"Oldest Movie Release Year: {df['release_year'].min()}")
      elif statistics == "2":
        print(f"Newest Movie Release Year: {df['release_year'].max()}")
      elif statistics == "3":
        print(f"Average Release Year: {df['release_year'].mean():.2f}") # Format to 2 decimal places
      elif statistics == "4":
        print(f"Total Number of Movies/TV Shows: {len(df)}")
      elif statistics == "5":
        print("Number of Movies and TV Shows:\n", df["type"].value_counts())
      else:
        print("Invalid input")
    elif customer == "3":
      search = input("Search movie: ")
      if search in df["title"].values:
        print(df[df["title"] == search])
      else:
        print("Movie not found")
    elif customer == "4":
      year_input = input("Enter a year: ")
      try:
        year = int(year_input)
        if year in df["release_year"].values:
          print(df[df["release_year"] == year])
        else:
          print("Year not found")
      except ValueError:
        print("Invalid year input. Please enter a number.")
    elif customer == "5":
      genre = input("Enter a genre: ")
      # Use str.contains for partial matching and handle NaN values
      matched_genres = df[df["listed_in"].astype(str).str.contains(genre, case=False, na=False)]
      if not matched_genres.empty:
        print(matched_genres)
      else:
        print("Genre not found")
    elif customer == "6":
      director = input("Enter a director: ")
      # Use str.contains for partial matching and handle NaN values
      matched_directors = df[df["director"].astype(str).str.contains(director, case=False, na=False)]
      if not matched_directors.empty:
        print(matched_directors)
      else:
        print("Director not found")
    elif customer == "7":
      country = input("Enter a country: ")
      # Use str.contains for partial matching and handle NaN values
      matched_countries = df[df["country"].astype(str).str.contains(country, case=False, na=False)]
      if not matched_countries.empty:
        print(matched_countries)
      else:
        print("Country not found")
    elif customer == "8":
      print(df.isnull().sum())
    elif customer == "9":
      print("Thank you for using our program")
      break
    else:
      print("Invalid input")
movie()
```
